similar types to NumPy types) of the same type and an associated array of data labels,

```python
# Define a dictionary with state names as keys and population numbers as values
sdata = {"Ohio": 35000, "Texas": 71000, "Oregon": 16000, "Utah": 5000}

# Create a pandas Series from the dictionary
# The dictionary keys become the index labels, and the values become the data
obj3 = pd.Series(sdata)

# Display the Series
obj3
```

```python
obj3.to_dict()
```

```python
# Define a list of states, including some not present in the original sdata dictionary
states = ["California", "Ohio", "Oregon", "Texas"]

# Create a new Series using sdata, but restrict/reorder it based on the states list
# For states not found in sdata (e.g., "California"), the value will be NaN (missing)
obj4 = pd.Series(sdata, index=states)

# Display the new Series
obj4
```

```python
# Check for missing (NaN) values in obj4
# Returns a boolean Series: True for missing values, False otherwise
pd.isna(obj4)
```

```python
# Check for non-missing (not NaN) values in obj4
# Returns a boolean Series: True where values are present, False where they are missing
pd.notna(obj4)
```

```python
# Use the Series method .isna() to check for missing (NaN) values in obj4
# This is equivalent to pd.isna(obj4)
obj4.isna()
```

```python
# Display the original Series created from the sdata dictionary
obj3
```

```python
# Display the Series obj4, which uses a custom index (including a state not in sdata)
obj4
```

```python
# Add the two Series together
# This performs element-wise addition based on matching index labels
# If a label is missing in either Series, the result will be NaN for that label
obj3 + obj4
```

```python
# Assign a name to the Series; this is used as the column header when converting to a DataFrame
obj4.name = "population"

# Assign a name to the index; this is useful for labeling the index in tables and plots
obj4.index.name = "state"

# Display the updated Series with named index and values
obj4
```

### DataFrame

```python
# Define a dictionary with column names as keys and lists of values
# Each list represents a column in the future DataFrame
data = {
    "state": ["Ohio", "Ohio", "Ohio", "Nevada", "Nevada", "Nevada"],  # State names
    "year": [2000, 2001, 2002, 2001, 2002, 2003],                      # Corresponding years
    "pop": [1.5, 1.7, 3.6, 2.4, 2.9, 3.2]                              # Population figures
}

# Create a DataFrame from the dictionary
# Each key becomes a column; the lists are treated as rows in that column
frame = pd.DataFrame(data)

frame
```

```python
frame.head()
```

```python
frame.tail()
```

```python
# Create a DataFrame from the 'data' dictionary
# Specify the order of the columns explicitly using the 'columns' parameter
pd.DataFrame(data, columns=["year", "state", "pop"])
```

```python
# Create a new DataFrame from the 'data' dictionary
# Specify column order and include an additional column "debt" that doesn't exist in the original data
frame2 = pd.DataFrame(data, columns=["year", "state", "pop", "debt"])

# Display the resulting DataFrame
frame2
```

```python
# Access the column labels (i.e., the names of all columns) of the DataFrame
frame2.columns
```

```python
# Access the "state" column using dictionary-style indexing (recommended when column names are dynamic or contain spaces)
frame2["state"]
```

```python
# Access the "year" column using attribute-style access (works only if the column name is a valid Python identifier)
frame2.year
```

```python
# Access the row at label/index 1 using label-based indexing with .loc
# This returns a Series representing the row with index label 1
frame2.loc[1]
```

```python
# Access the row at position 2 (third row) using integer-based indexing with .iloc
# This also returns a Series representing the row at that numeric position
frame2.iloc[2]
```

```python
# Assign the scalar value 16.5 to every row in the "debt" column
# Since "debt" previously contained NaNs, this replaces them with 16.5
frame2["debt"] = 16.5

# Display the updated DataFrame with all debt values set to 16.5
frame2
```

```python
# Replace the "debt" column with an array of 6 values: [0., 1., 2., 3., 4., 5.]
# np.arange(6.) creates a NumPy array of length 6 with float values from 0 to 5
# The array aligns by position with the rows in the DataFrame
frame2["debt"] = np.arange(6.)

# Display the updated DataFrame with the new debt values
frame2
```

```python
# Create a Series named 'val' with custom index labels
# These labels correspond to index values in frame2
val = pd.Series([-1.2, -1.5, -1.7], index=["two", "four", "five"])

# Assign the Series 'val' to the "debt" column in frame2
# Pandas aligns values by index label, so only rows with matching index labels ("two", "four", "five")
# will receive values; other rows will get NaN
frame2["debt"] = val

# Display the updated DataFrame
frame2
```

```python
# Create a new column "eastern" in frame2
# This column contains boolean values: True where the "state" is "Ohio", False otherwise
frame2["eastern"] = frame2["state"] == "Ohio"

# Display the updated DataFrame with the new "eastern" column
frame2
```

```python
# Delete the "eastern" column from frame2 (if it exists)
# If "eastern" doesn't exist, this will raise a KeyError
del frame2["eastern"]

# View the current list of column names in frame2 after deletion
frame2.columns
```

### Reindexing

```python
# Create a pandas Series with values [4.5, 7.2, -5.3, 3.6]
# Assign custom index labels ["d", "b", "a", "c"] to these values
obj = pd.Series([4.5, 7.2, -5.3, 3.6], index=["d", "b", "a", "c"])

# Display the Series
obj
```

```python
# Reindex the Series 'obj' to include a new set of index labels: ["a", "b", "c", "d", "e"]
# This reorders existing entries and introduces a new label "e", which will have a NaN value
obj2 = obj.reindex(["a", "b", "c", "d", "e"])

# Display the resulting Series
obj2
```

```python
# Create a Series with string values and a non-continuous integer index
obj3 = pd.Series(["blue", "purple", "yellow"], index=[0, 2, 4])

# Display the Series
obj3
```

```python
# Reindex obj3 to include a full range of indices from 0 to 5
# Use forward-fill ('ffill') to propagate the last valid value forward into missing entries
obj3.reindex(np.arange(6), method="ffill")
```

```python
# Create a 3x3 DataFrame with values from 0 to 8, reshaped as a 3-row matrix
# Assign custom row index labels: ["a", "c", "d"]
# Assign column names: ["Ohio", "Texas", "California"]
frame = pd.DataFrame(np.arange(9).reshape((3, 3)),
                     index=["a", "c", "d"],
                     columns=["Ohio", "Texas", "California"])

# Display the original DataFrame
frame
```

```python
# Reindex the DataFrame to include a new row label "b"
# Existing rows are retained, and "b" is added with NaNs for all columns (since no data existed for it)
frame2 = frame.reindex(index=["a", "b", "c", "d"])

# Display the reindexed DataFrame
frame2
```

```python
# Define a list of column names to reindex the DataFrame's columns
# Includes one column ("Utah") that doesn't exist in the original DataFrame
states = ["Texas", "Utah", "California"]

# Reindex the DataFrame 'frame' to include only the specified columns, in the given order
# Columns not present in the original (like "Utah") will be filled with NaN
frame.reindex(columns=states)
```

```python
# Reindex the columns of 'frame' using the list 'states'
# Equivalent to frame.reindex(columns=states), just a different syntax
# axis="columns" (or axis=1) specifies you're reindexing the columns, not the rows
frame.reindex(states, axis="columns")
```

```python
# Use .loc to select specific rows and columns from the DataFrame 'frame'
# Row labels: ["a", "d", "c"]
# Column labels: ["California", "Texas"]
frame.loc[["a", "d", "c"], ["California", "Texas"]]
```

```python
print(obj1)
print(obj2)
```

```python
# For obj1, the index labels are [2, 0, 1]
# When you do obj1[[0, 1, 2]], you're selecting by **label**, not position
# So this tries to retrieve index labels 0, 1, and 2 — which all exist
obj1[[0, 1, 2]]  # Returns values in label order: 0 → 1 → 2
```

```python
# For obj2, the index labels are ["a", "b", "c"]
# obj2[[0, 1, 2]] will raise a KeyError, because labels 0, 1, and 2 do not exist
# It is trying to access labels 0, 1, 2 — not positions
obj2[[0, 1, 2]]  # ❌ KeyError
```

```python
# Use .loc to perform **label-based slicing** from "b" to "c" (inclusive)
# Since Series index is ["a", "b", "c"], this returns elements with labels "b" and "c"
obj2.loc["b":"c"]
```

```python
# Set the values of the index labels from "b" to "c" (inclusive) to 5
# Since obj2 has index ["a", "b", "c"], this will update both "b" and "c"
obj2.loc["b":"c"] = 5

# Display the modified Series
obj2
```

```python
# Create a 4x3 DataFrame `frame` filled with random values 
# drawn from a standard normal distribution (mean=0, std=1)
# Columns are labeled "b", "d", and "e"
# Rows are labeled by state names
frame = pd.DataFrame(np.random.standard_normal((4, 3)),
                     columns=list("bde"),
                     index=["Utah", "Ohio", "Texas", "Oregon"])

# Display the randomly generated DataFrame
frame
```

```python
# Define a function that computes the range (max - min) of a Series
# This function will be applied to each column (by default) of the DataFrame
def f1(x):
    return x.max() - x.min()

# Apply the function `f1` to each column of the DataFrame `frame`
# This returns a Series containing the range for each column
frame.apply(f1)
```

```python
# Apply the custom function `f1` to each row of the DataFrame
# `f1` computes the difference between the max and min of a Series
# Setting axis="columns" applies the function across columns (i.e., row-wise)
frame.apply(f1, axis="columns")
```

```python
# Define a function that returns the min and max of a Series as a new Series
def f2(x):
    return pd.Series([x.min(), x.max()], index=["min", "max"])

# Apply the function `f2` to each column of the DataFrame (default axis=0)
# This returns a DataFrame where each column contains the min and max values of the original columns
frame.apply(f2)
```

### Unique Values, Value Counts, and Membership

```python
# Create a pandas Series with some repeated string values
obj = pd.Series(["c", "a", "d", "a", "a", "b", "b", "c", "c"])
```

```python
# Get the unique values in the Series
uniques = obj.unique()
uniques  # Output: array of unique elements from the Series
```

```python
# Count the frequency of each unique value in the Series
obj.value_counts()
# Output: Series showing unique values as index and their counts as values, sorted in descending order
```

```python
# Count unique values in the Series after converting to a NumPy array,
# without sorting the result
pd.value_counts(obj.to_numpy(), sort=False)
# Output: Series showing counts of each unique value in the original order of appearance
```

```python
obj
# Displays the original Series

mask = obj.isin(["b", "c"])
# Creates a boolean mask identifying elements that are either 'b' or 'c'

mask
# Displays the boolean mask

obj[mask]
# Filters the Series using the mask to show only the elements that are 'b' or 'c'
```

```python
to_match = pd.Series(["c", "a", "b", "b", "c", "a"])
# Series of values that we want to match against unique values

unique_vals = pd.Series(["c", "b", "a"])
# Series of unique values to match from

indices = pd.Index(unique_vals).get_indexer(to_match)
# Uses get_indexer to find the indices of each element in to_match within unique_vals
# Returns -1 if an element in to_match is not found in unique_vals

indices
# Displays the resulting index positions for each element in to_match
```
